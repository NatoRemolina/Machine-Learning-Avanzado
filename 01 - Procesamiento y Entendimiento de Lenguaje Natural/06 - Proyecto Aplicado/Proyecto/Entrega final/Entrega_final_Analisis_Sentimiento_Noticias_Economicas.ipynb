#### **2.1.2.1. Valores nulos, documentos vacíos y duplicados**

`isnull()` permite detectar valores ausentes, pero no identifica cadenas vacías como `""`
o `"   "`. Por ello se verifican ambas situaciones explícitamente.


```python
# Auditoría básica de calidad del corpus español
quality_basic = pd.DataFrame(index=es_data.columns)

quality_basic['nulos'] = es_data.isna().sum()
quality_basic['porcentaje_nulos'] = (
    100 * quality_basic['nulos'] / len(es_data)
).round(2)

# Solo tiene sentido buscar cadenas vacías en columnas de texto.
for col in ['sentence', 'label', 'source', 'language']:
    s = es_data[col].fillna('').astype(str)
    quality_basic.loc[col, 'vacios_o_espacios'] = int(s.str.strip().eq('').sum())

display(quality_basic)

print('Duplicados exactos de sentence:',
      int(es_data.duplicated(subset=['sentence']).sum()))
print('Filas completamente duplicadas:',
      int(es_data.duplicated().sum()))
```

#### **2.1.2.3. Indicios de problemas de codificación**

Los caracteres Unicode válidos no son necesariamente errores. Por lo cual:

Se buscan:

- `�`: carácter de reemplazo Unicode, indicador fuerte de pérdida durante una decodificación;
- `Ã`, `Â`, `â`: patrones comunes de *mojibake*;
- `ń`: carácter válido en otros idiomas, pero atípico en español y observado en algunos
  fragmentos del corpus;
- caracteres de control no imprimibles;
- posibles etiquetas HTML residuales.

```python
sentence_text = es_data['sentence'].fillna('').astype(str)

encoding_checks = pd.Series({
    'caracter_reemplazo_�': sentence_text.str.contains('\uFFFD', regex=False).sum(),
    'posible_mojibake_Ã': sentence_text.str.contains('Ã', regex=False).sum(),
    'posible_mojibake_Â': sentence_text.str.contains('Â', regex=False).sum(),
    'posible_mojibake_â': sentence_text.str.contains('â', regex=False).sum(),
    'caracter_atipico_ń': sentence_text.str.contains('ń', regex=False).sum(),
    'caracteres_control': sentence_text.str.contains(
        r'[\x00-\x08\x0B\x0C\x0E-\x1F]', regex=True
    ).sum(),
    'posible_html': sentence_text.str.contains(r'<[^>]+>', regex=True).sum()
}).astype(int)

display(encoding_checks.to_frame('conteo'))

encoding_pattern = r'�|Ã|Â|â|ń|[\x00-\x08\x0B\x0C\x0E-\x1F]|<[^>]+>'
encoding_flag = sentence_text.str.contains(encoding_pattern, regex=True)

print('Documentos con alguna bandera de codificación/grafía:',
      int(encoding_flag.sum()))

if encoding_flag.any():
    display(
        es_data.loc[encoding_flag, ['sentence', 'label', 'source']]
        .head(10)
    )
```

#### **2.1.2.4. Legibilidad y características de calidad**

Se conserva el índice de Fernández-Huerta como una descripción de complejidad lingüística.
No se interpreta como una medida de "veracidad" ni como un filtro de sentimiento.

Además se calcula una medida simple de proporción de tokens alfabéticos para detectar
fragmentos dominados por símbolos o ruido. Esta métrica tampoco elimina registros por sí sola;
sirve para inspeccionar extremos.


```python
def alphabetic_token_ratio(text):
    doc = nlp(str(text))
    tokens = [t for t in doc if not t.is_space]
    if not tokens:
        return 0.0
    return np.mean([t.is_alpha or t.like_num for t in tokens])

# Métricas descriptivas de calidad
reading_quality = es_data['sentence'].astype(str).map(textstat.fernandez_huerta)
alpha_token_ratio = es_data['sentence'].astype(str).map(alphabetic_token_ratio)

quality_metrics = pd.DataFrame({
    'fernandez_huerta': reading_quality,
    'proporcion_tokens_alfanumericos': alpha_token_ratio
})

display(
    quality_metrics.describe(
        percentiles=[0.05, 0.25, 0.50, 0.75, 0.95]
    ).T
)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].hist(reading_quality.dropna(), bins=40)
axes[0].set_title('Índice de Fernández-Huerta')
axes[0].set_xlabel('Puntaje')
axes[0].set_ylabel('Número de documentos')

axes[1].hist(alpha_token_ratio.dropna(), bins=40)
axes[1].set_title('Proporción de tokens alfanuméricos')
axes[1].set_xlabel('Proporción')
axes[1].set_ylabel('Número de documentos')

plt.tight_layout()
plt.show()
```

Los documentos con indicios de codificación se **conservan**, pero se marcan para que el
preprocesamiento pueda normalizar lo que sea recuperable sin inventar caracteres.


```python
# Máscaras de calidad reproducibles
sentence_clean_mask = (
    es_data['sentence'].notna()
    & es_data['sentence'].astype(str).str.strip().ne('')
)

label_clean_mask = (
    es_data['label']
    .fillna('')
    .astype(str)
    .str.strip()
    .str.lower()
    .isin(expected_labels)
)

language_clean_mask = es_data['language'].eq('es')

data_corrected = (
    es_data.loc[
        sentence_clean_mask & label_clean_mask & language_clean_mask
    ]
    .copy()
)

# Normalización mínima de las etiquetas antes de continuar.
data_corrected['label'] = (
    data_corrected['label']
    .astype(str)
    .str.strip()
    .str.lower()
)

before_dedup = len(data_corrected)
data_corrected = data_corrected.drop_duplicates(
    subset=['sentence'],
    keep='first'
).copy()

removed_duplicates = before_dedup - len(data_corrected)

# Bandera informativa; no elimina documentos.
data_corrected['encoding_flag'] = (
    data_corrected['sentence']
    .fillna('')
    .astype(str)
    .str.contains(encoding_pattern, regex=True)
)

quality_filter_summary = pd.DataFrame({
    'indicador': [
        'Documentos españoles iniciales',
        'Documentos válidos después de filtros objetivos',
        'Duplicados eliminados',
        'Documentos conservados con bandera de codificación',
        'Porcentaje de conservación'
    ],
    'valor': [
        len(es_data),
        len(data_corrected),
        removed_duplicates,
        int(data_corrected['encoding_flag'].sum()),
        round(100 * len(data_corrected) / len(es_data), 2)
    ]
})

display(quality_filter_summary)
```

Como se puede apreciar en las etapas anteriores el corpus que continúa en el EDA evita eliminar noticias por
complejidad lingüística. De esta forma se reduce el riesgo de alterar artificialmente la
distribución de fuentes o sentimientos. Las métricas de legibilidad se conservan como
caracterización descriptiva y los posibles artefactos de codificación quedan señalados
para el preprocesamiento.

#### **2.1.3.2. Longitud y diversidad léxica de los documentos**

Se caracterizan los textos mediante:

- número de palabras;
- *Type-Token Ratio* (TTR), proporción de palabras únicas respecto al total.

El TTR se interpreta con cautela porque depende de la longitud: textos cortos tienden a
obtener valores altos con mayor facilidad.


```python
text_length_chars = data_corrected['sentence'].astype(str).map(len)
text_length_words = data_corrected['sentence'].astype(str).map(
    lambda x: len(x.split())
)

def type_token_ratio(text):
    tokens = [t.lower() for t in str(text).split() if t.strip()]
    return len(set(tokens)) / len(tokens) if tokens else 0.0

text_ttr = data_corrected['sentence'].map(type_token_ratio)

text_stats = pd.DataFrame({
    'caracteres': text_length_chars,
    'palabras': text_length_words,
    'TTR': text_ttr
}, index=data_corrected.index)

display(
    text_stats.describe(
        percentiles=[0.05, 0.25, 0.50, 0.75, 0.95]
    ).T
)

fig, axes = plt.subplots(1, 3, figsize=(15, 5))

axes[0].hist(text_stats['caracteres'], bins=40)
axes[0].set_title('Longitud en caracteres')
axes[0].set_xlabel('Caracteres')
axes[0].set_ylabel('Número de documentos')

axes[1].hist(text_stats['palabras'], bins=40)
axes[1].set_title('Longitud en palabras')
axes[1].set_xlabel('Palabras')
axes[1].set_ylabel('Número de documentos')

axes[2].hist(text_stats['TTR'], bins=40)
axes[2].set_title('Type-Token Ratio')
axes[2].set_xlabel('TTR')
axes[2].set_ylabel('Número de documentos')

plt.tight_layout()
plt.show()
```

### **2.1.4. Relación Entre Variables y Redundancia**
---

Como el dataset contiene principalmente texto y variables categóricas, una matriz de
correlación de Pearson sobre las columnas originales no sería apropiada.

En su lugar se estudian relaciones que sí tienen interpretación:

1. `source` frente a `label`, mediante una tabla de contingencia y el **V de Cramér**;
2. longitud/TTR frente a `label`, mediante estadísticas y diagramas de caja;
3. redundancia entre las medidas numéricas derivadas del texto mediante correlación de Spearman;
4. redundancia estructural de `language`, que es constante una vez filtrado el español.


```python
# Asociación entre source y label
from scipy.stats import chi2_contingency

source_label_table = pd.crosstab(
    data_corrected['source'],
    data_corrected['label']
)

source_label_pct = pd.crosstab(
    data_corrected['source'],
    data_corrected['label'],
    normalize='index'
) * 100

# V de Cramér
chi2, p_value, _, _ = chi2_contingency(source_label_table)
n = source_label_table.to_numpy().sum()
min_dim = min(source_label_table.shape) - 1
cramers_v = np.sqrt((chi2 / n) / min_dim) if min_dim > 0 else np.nan

print(f'V de Cramér(source, label): {cramers_v:.3f}')
print(f'p-value del contraste chi-cuadrado: {p_value:.3e}')

# Visualizar solo las fuentes más representadas para mantener legibilidad
top_source_names = (
    data_corrected['source']
    .value_counts()
    .head(12)
    .index
)

heatmap_data = source_label_pct.loc[
    source_label_pct.index.intersection(top_source_names)
]

plt.figure(figsize=(9, max(5, 0.45 * len(heatmap_data))))
sns.heatmap(
    heatmap_data[['negative', 'neutral', 'positive']],
    annot=True,
    fmt='.1f',
    cmap='Blues'
)
plt.title('Distribución porcentual del sentimiento por fuente')
plt.xlabel('Sentimiento')
plt.ylabel('Fuente')
plt.tight_layout()
plt.show()
```

El V de Cramér resume la intensidad de la asociación entre dos variables categóricas.
El resultado debe interpretarse junto con la tabla y el tamaño de muestra. Una asociación
entre `source` y `label` no implica causalidad, pero sí puede advertir que el corpus contiene
diferencias de distribución entre fuentes.

Esto justifica mantener `source` como variable de **auditoría** y no incluirla de forma
automática como predictor en una primera línea base.

## **2.2. Limpieza de los Textos**
---

El preprocesamiento debe ser coherente con el objetivo de **clasificar sentimiento en noticias
económicas**. Por esta razón no conviene eliminar indiscriminadamente toda la información que
parece "ruido":

- las **negaciones** (`no`, `nunca`, `sin`, etc.) pueden cambiar el sentimiento;
- los **números, porcentajes y monedas** son informativos en noticias financieras;
- los acentos y caracteres españoles son información lingüística válida.

La estrategia aplicada combina normalización Unicode, tokenización, lematización, filtrado
conservador de *stop words* y preservación de señales financieras mediante marcadores.


### **2.2.1. Decisiones de preprocesamiento**

Se aplicarán las siguientes transformaciones:

1. **Normalización Unicode NFKC** y eliminación de caracteres de control invisibles.
2. Sustitución del carácter de reemplazo `�` por espacio. No se intenta adivinar el carácter
   perdido.
3. **Tokenización con spaCy**.
4. **Lematización** de palabras alfabéticas.
5. **Minúsculas** manteniendo caracteres propios del español.
6. Eliminación de puntuación y espacios redundantes.
7. Eliminación de *stop words* de forma conservadora.
8. **Preservación de negaciones** aunque spaCy las marque como *stop words*.
9. Normalización de números, porcentajes y monedas mediante los tokens `num`, `pct` y `moneda`.
10. Conservación del texto original en `sentence_original`; el resultado procesado se guarda
    en una nueva columna `sentence_clean`.

No se realiza corrección ortográfica automática, ya que una corrección no supervisada podría
alterar nombres de empresas, productos o entidades económicas.


```python
import unicodedata

# Negaciones relevantes para análisis de sentimiento que NO se eliminan.
NEGATIONS_TO_KEEP = {
    'no', 'nunca', 'jamás', 'jamas', 'sin', 'ni', 'tampoco', 'nadie', 'ninguno', 'ninguna'
}

CURRENCY_SYMBOLS = {'$', '€', '£', '¥', '₩', '₹', '₽', '₺'}

def normalize_unicode_text(text):
    text = unicodedata.normalize('NFKC', str(text))
    text = text.replace('\uFFFD', ' ')   # carácter de reemplazo
    text = text.replace('\u200b', ' ')   # zero-width space
    text = text.replace('\ufeff', ' ')   # BOM / zero-width no-break
    text = re.sub(r'\s+', ' ', text)
    return text.strip()

def preprocess_doc(doc):
    cleaned_tokens = []

    for token in doc:
        raw = token.text.strip()
        lower = raw.lower()

        if not raw or token.is_space:
            continue

        # Señales financieras: se conservan como marcadores semánticos.
        if token.like_num:
            cleaned_tokens.append('num')
            continue

        if raw in {'%', '％'}:
            cleaned_tokens.append('pct')
            continue

        if raw in CURRENCY_SYMBOLS:
            cleaned_tokens.append('moneda')
            continue

        # Se descarta puntuación y símbolos no informativos.
        if token.is_punct:
            continue

        # Solo se conservan palabras alfabéticas en este punto.
        if not token.is_alpha:
            continue

        # Las negaciones se conservan aunque sean stop words.
        if token.is_stop and lower not in NEGATIONS_TO_KEEP:
            continue

        lemma = token.lemma_.lower().strip()
        lemma = unicodedata.normalize('NFKC', lemma)

        if lemma:
            cleaned_tokens.append(lemma)

    return ' '.join(cleaned_tokens)

def preprocess_series(series, batch_size=256):
    normalized = series.fillna('').astype(str).map(normalize_unicode_text)

    disabled = [
        pipe_name
        for pipe_name in ['parser', 'ner', 'language_detector']
        if pipe_name in nlp.pipe_names
    ]

    docs = nlp.pipe(
        normalized.tolist(),
        batch_size=batch_size,
        disable=disabled
    )

    return [preprocess_doc(doc) for doc in docs]
```

### **2.2.2. Aplicación del preprocesamiento**

En esta seccion el texto original no se sobrescribe, lo cual permite auditar cualquier cambio y comparar antes
y después. El corpus final conserva también `label`, `source`, `language` y la bandera de
codificación generada en el punto 1.


```python
data_clean = data_corrected.copy()

data_clean['sentence_original'] = data_clean['sentence']
data_clean['sentence_clean'] = preprocess_series(data_clean['sentence'])

# Controles posteriores
empty_after = data_clean['sentence_clean'].fillna('').str.strip().eq('')

print('Documentos antes del preprocesamiento:', len(data_clean))
print('Documentos vacíos después del preprocesamiento:', int(empty_after.sum()))

# Si alguna fila pierde todo su contenido semántico, se elimina y se documenta.
if empty_after.any():
    removed_after_preprocess = int(empty_after.sum())
    data_clean = data_clean.loc[~empty_after].copy()
else:
    removed_after_preprocess = 0

print('Documentos eliminados por quedar vacíos:', removed_after_preprocess)
print('Documentos finales:', len(data_clean))

display(
    data_clean[
        ['sentence_original', 'sentence_clean', 'label', 'source']
    ].head()
)
```

## **3.1. Representación Numérica del Texto: Embeddings**

En este análisis, nos centraremos en el texto contenido en la columna `sentence_clean` de nuestro conjunto de datos. El objetivo principal es comprender la necesidad y las diferentes técnicas para representar texto de forma numérica, un paso crucial en el procesamiento del lenguaje natural (NLP) para el aprendizaje automático.

Cuando trabajamos con texto como variable de entrada en modelos de aprendizaje automático, surge una limitación fundamental: las computadoras no "entienden" el texto en su forma natural. Los algoritmos de machine learning operan con datos numéricos, realizando cálculos matemáticos sobre ellos. Por lo tanto, no podemos alimentar directamente los modelos con cadenas de texto.

Para superar esta barrera, necesitamos transformar el texto en una representación numérica que capture su significado semántico y sintáctico, permitiendo que los algoritmos puedan procesarlo y aprender de él. Estas representaciones se conocen comúnmente como *embeddings* (incrustaciones). Un embedding representa una palabra, frase o documento como un vector numérico en un espacio multidimensional. La posición de este vector en el espacio refleja el significado del elemento representado: palabras con significados similares estarán ubicadas cerca unas de otras.

A continuación realizaremos el ajuste de *Word2vec* y *Doc2vec*. Les recordamos que este segundo no es recomendable debido al volumen de nuestros datos. Sin embargo, validaremos dicha hipótesis.

Por otra parte, es importante tener en cuenta que los métodos profundos carecen de interpretabilidad, siendo propicios para mejorar el desempeño general de modelos de machine learning, reduciendo la capacidad de extraer cocimiento más allá de la capacidad predictiva del modelo.

#### **3.3.3.1. Word2Vec**

```python
logging.disable(logging.CRITICAL)

# En gensim 'workers' es el número de hilos de entrenamiento: con workers = -1 no se crea
# ningún hilo y el modelo queda sin entrenar. Se usa el número de núcleos disponibles.
N_WORKERS = os.cpu_count()

tokens = list(map(lambda doc: doc.split(), X_train))
word_model = Word2Vec(
        sentences = tokens,
        vector_size = 512,
        window = 5,
        min_count = 2,
        epochs = 20,
        seed = 2402,
        workers = N_WORKERS)

print(f'Tamaño del vocabulario aprendido: {len(word_model.wv.key_to_index)} palabras')
```

Recuerde, como se señaló anteriormente, que en este caso el modelo extraerá una representación para cada palabra y cada oración será representada por el vector promedio entre las palabras que lo componen.


```python
def vector_documento(doc, modelo, dim=512):
    """Representa una oración como el PROMEDIO de los vectores de las palabras que la componen.

    Se usa el promedio y no la suma porque la suma hace que la magnitud del vector crezca
    con la longitud del texto, midiendo longitud en lugar de contenido semántico.
    """
    vectores = [modelo.wv[token] for token in doc.split() if token in modelo.wv.key_to_index]
    return np.mean(vectores, axis=0) if vectores else np.zeros(dim)

word2vect_embedding = pd.DataFrame([vector_documento(doc, word_model) for doc in X_tune])
word2vect_embedding.head()
```

#### **3.3.3.2. Doc2Vec**

```python
logging.disable(logging.CRITICAL)

tokens = list(map(lambda doc: doc.split(), X_train))
tagged_corpus = [TaggedDocument(doc, [i]) for i, doc in enumerate(tokens)]

model_trained = Doc2Vec(
    documents = tagged_corpus,
    vector_size = 512,
    window = 5,
    min_count = 2,
    epochs = 20,
    seed = 2402,
    workers = N_WORKERS   # mismo ajuste que en Word2Vec
)
```

```python
tokens = list(map(lambda doc: doc.split(), X_tune))
tagged_corpus = [TaggedDocument(doc, [i]) for i, doc in enumerate(tokens)]
doc2vect_embedding = []

for i, doc in enumerate(tagged_corpus):
  vect = model_trained.infer_vector(doc.words)
  doc2vect_embedding.append(vect)
  #print(f'{i} of {len(tagged_corpus)}', end='\r')

doc2vect_embedding = pd.DataFrame(doc2vect_embedding)
doc2vect_embedding.head()
```

## **3.4. Exploración del Embedding**
---

Una vez que hemos ajustado y extraído las representaciones del conjunto de ajuste de hiperparámetros *tune set*, necesitamos evaluar la calidad de estas. Esto lo realizaremos bajo tres criterios complementarios:

1. **Redundancia interna:** qué tanta información repetida contienen las características extraídas.
2. **Asociación con la variable objetivo:** qué tan relacionada está cada característica con `label`.
3. **Verificación multivariada:** si la señal detectada es efectivamente aprovechable por un clasificador.

Para poder calcular correlaciones, la variable objetivo se codifica como ordinal, dado que `label` es una variable de tipo texto (`negative`, `neutral`, `positive`).

#### **3.4.1. Redundancia interna**

La redundancia se refiere al fenómeno en el que la mayoría de las características de nuestro conjunto de datos están correlacionadas entre sí. Generalmente ocurre cuando los descriptores extraídos no siguen un patrón regular, lo que hace que su distribución sea aleatoria. Una representación redundante desperdicia dimensiones: aunque tengamos 512 características, la información contenida es mucho menor.

Para validarlo analizaremos los mapas de correlación y, adicionalmente, resumiremos la magnitud de las correlaciones fuera de la diagonal, ya que con 512 características el mapa de calor por sí solo no permite cuantificar el fenómeno.

```python
# Diccionario con las cuatro representaciones extraídas sobre el tune set
embeddings = {
    'BoW':      Bow_embeddings,
    'TF-IDF':   TFIDFembedding,
    'Word2Vec': word2vect_embedding,
    'Doc2Vec':  doc2vect_embedding,
}

# La variable objetivo es categórica (negative / neutral / positive): para calcular
# correlaciones se codifica como ordinal negative < neutral < positive.
orden_label = {'negative': 0, 'neutral': 1, 'positive': 2}
y_tune_num = y_tune.map(orden_label).astype(float).reset_index(drop=True)
y_tune_cat = y_tune.reset_index(drop=True)

print(y_tune_cat.value_counts())
print(f'Registros en tune: {len(y_tune_num)} | nulos tras codificar: {y_tune_num.isna().sum()}')

def matriz_correlacion(df):
    """Correlación de Pearson descartando características constantes (sin varianza)."""
    df_valido = df.loc[:, df.std(numeric_only=True) > 0]
    return df_valido.corr()

def resumen_redundancia(corr):
    """Estadísticos de las correlaciones fuera de la diagonal (triángulo superior)."""
    valores = corr.abs().values
    iu = np.triu_indices_from(valores, k=1)
    fuera_diagonal = valores[iu]
    fuera_diagonal = fuera_diagonal[~np.isnan(fuera_diagonal)]
    return pd.Series({
        'n_caracteristicas': corr.shape[1],
        '|r| medio':         fuera_diagonal.mean(),
        '|r| percentil 95':  np.percentile(fuera_diagonal, 95),
        '|r| maximo':        fuera_diagonal.max(),
        '% pares |r| > 0.8': (fuera_diagonal > 0.8).mean() * 100,
    })

matrices = {nombre: matriz_correlacion(df) for nombre, df in embeddings.items()}

fig, axes = plt.subplots(1, 4, figsize=(22, 5))
for ax, (nombre, corr) in zip(axes, matrices.items()):
    sns.heatmap(corr, cmap='coolwarm', vmin=-1, vmax=1, ax=ax,
                xticklabels=False, yticklabels=False, cbar_kws={'shrink': 0.7})
    ax.set_title(f'Mapa de correlaciones {nombre}')

plt.tight_layout()
plt.show()

tabla_redundancia = pd.DataFrame({n: resumen_redundancia(m) for n, m in matrices.items()}).T
tabla_redundancia.round(4)
```

Los métodos frecuentistas (**BoW** y **TF-IDF**) presentan correlaciones fuera de la diagonal muy bajas (|r| medio cercano a 0.02), lo que indica que cada característica aporta información propia: no existe redundancia.

En contraste, **Word2Vec promediado resulta altamente redundante** (|r| medio cercano a 0.51 y aproximadamente un 18% de los pares con |r| > 0.8). Esto ocurre porque, al promediar los vectores de palabra, las 512 dimensiones colapsan sobre un subespacio de dimensionalidad efectiva mucho menor. **Doc2Vec** se ubica en un punto intermedio.

El valor máximo cercano a 1 que aparece en BoW y TF-IDF corresponde a un par aislado de términos que siempre concurren y no a un patrón general, como lo confirma el percentil 95 de las correlaciones.

Conclusión de este criterio: las representaciones frecuentistas aprovechan mejor las 512 dimensiones disponibles, mientras que las profundas requerirían regularización o reducción de dimensionalidad en la etapa de modelamiento.

#### **4.2.2.1. Word2Vec Wrapper**

Es fundamental recordar que **Word2Vec** no está implementado de forma nativa en scikit-learn (sklearn), lo que dificulta su integración directa con los modelos de esta librería. Para facilitar este proceso, es necesario crear una función *wrapper* o envoltorio. Esta función actúa como un puente, adaptando y reestructurando los objetos y salidas generadas por gensim para que sean compatibles y comprensibles para sklearn.

```python
class Word2VecWrapper(BaseEstimator):
    def __init__(self, max_features=100, epochs=10):
        self.max_features = max_features
        self.epochs = epochs
        self.model = None

    def fit(self, X, y=None):
        tokens = list(map(lambda doc: doc.split(), X))
        self.model = Word2Vec(sentences = tokens, vector_size = self.max_features, epochs = self.epochs, workers = -1)
        return self

    def transform(self, X):
        word2vect_embedding = []
        for i, doc in enumerate(X):
          doc_tokens = doc.split()
          vect = np.zeros(self.max_features)
          for token in doc_tokens:
            if token in self.model.wv.key_to_index:
              vect += self.model.wv[token]

          word2vect_embedding.append(vect)
        return np.array(word2vect_embedding)
```

Teniendo adaptado nuestros métodos de representación, debemos seleccionar los modelos de aprendizaje. Dado que nos enfrentamos a un problema de clasificación, exploraremos tres opciones que abarcan diferentes enfoques y capacidades:

1.  **Regresión Logística:** Este modelo, a pesar de su simplicidad, es una excelente opción inicial para determinar si el problema es linealmente separable. Su interpretabilidad nos permite comprender qué características son más relevantes para la clasificación.
2.  **Máquinas de Soporte Vectorial (SVM):** En caso de que la regresión logística no ofrezca resultados satisfactorios debido a la no linealidad del problema, las SVM nos permiten explorar soluciones más complejas. Mediante el uso de kernels, como el RBF (Radial Basis Function), podemos mapear los datos a un espacio de mayor dimensión donde las clases sean linealmente separables.
3. **Bosques Aleatorios (Random Forests):** Los bosques aleatorios son una alternativa popular y robusta, especialmente para datos tabulares como la representación extraída por TF-IDF. Su capacidad para manejar datos de alta dimensionalidad y su menor susceptibilidad al sobreajuste los convierten en una opción atractiva para este tipo de problemas.

La evaluación de estos tres modelos nos permitirá determinar cuál de ellos se adapta mejor a nuestros datos y objetivos, considerando tanto su rendimiento predictivo como su interpretabilidad y capacidad para generalizar a datos no vistos.

Por otra parte, como se observa en la distribución de la variable objetivo, existe un desbalanceo significativo entre las clases, lo que podría afectar el rendimiento del modelo al favorecer la clase mayoritaria. Para mitigar este problema, se ha configurado el parámetro `class_weight = "balanced"` en todos los modelos utilizados. Esta estrategia ajusta los pesos de cada clase de forma inversamente proporcional a su frecuencia, otorgando mayor importancia a las clases minoritarias durante el entrenamiento. Al hacerlo, se reduce el efecto del desbalanceo y se busca que el modelo aprenda a predecir de manera más equitativa todas las clases, mejorando su capacidad de generalización y evitando un sesgo hacia la clase dominante.

```python
#Pipeline regresión logistica - TF-IDF
pipeline_lr_TFIDF = Pipeline([('vectorizer', TfidfVectorizer()),
                              ('classifier', LogisticRegression(class_weight="balanced"))])

#Pipeline SVM  - TF-IDF
pipeline_svm_TFIDF = Pipeline([('vectorizer', TfidfVectorizer()),
                               ('classifier', SVC(kernel='rbf', class_weight="balanced", max_iter=50))])

#Pipeline random forest  - TF-IDF
pipeline_rf_TFIDF = Pipeline([('vectorizer', TfidfVectorizer()),
                        ('classifier', RandomForestClassifier(class_weight="balanced", max_samples= 100))])

#Pipeline regresión logistica - Word2Vec
pipeline_lr_W2V = Pipeline([('vectorizer', Word2VecWrapper()),
                        ('classifier', LogisticRegression(class_weight="balanced"))])

#Pipeline SVM - Word2Vec
pipeline_svm_W2V = Pipeline([('vectorizer', Word2VecWrapper()),
                             ('classifier', SVC(kernel='rbf', class_weight="balanced", max_iter=50))])

#Pipeline random forest - Word2Vec
pipeline_rf_W2V = Pipeline([('vectorizer', Word2VecWrapper()),
                            ('classifier', RandomForestClassifier(class_weight="balanced", max_samples= 100))])
```

Una vez generado los flujos de trabajo "**pipelines**" es necesario establecer el espacio de hiperparámetros para cada uno. Recuerde que los hiperparámetros son factores que deben proporcionarse inicialmente al modelo y que pueden afectar considerablemente su desempeño. En la siguiente tabla encontrará los hiperparámetros identificados para cada modelo y su espacio de búsqueda.

Una vez generados los flujos de trabajo ("pipelines"), es fundamental definir el espacio de hiperparámetros para cada modelo. Recordemos que los hiperparámetros son parámetros que se configuran antes del entrenamiento y que pueden influir significativamente en el desempeño del modelo. La selección adecuada del espacio de búsqueda es crucial para encontrar la combinación óptima de hiperparámetros que maximice el rendimiento del modelo.

A continuación, se presenta una tabla con los hiperparámetros identificados para cada modelo y su correspondiente espacio de búsqueda. Es importante destacar que este espacio ha sido definido considerando el conocimiento previo sobre el problema en cuestión. Se han incluido valores que abarcan un rango amplio de posibilidades, permitiendo explorar diferentes configuraciones y evaluar su impacto en el rendimiento del modelo.

La exploración exhaustiva de este espacio de hiperparámetros se llevará a cabo mediante la técnica de optimización conocida como la búsqueda en malla (**grid search**). Estas técnicas nos permitirán identificar la combinación de hiperparámetros que mejor se adapte a nuestros datos y objetivos, garantizando que el modelo final tenga un desempeño óptimo en la tarea de clasificación.



```python
lr_param_grid = {'vectorizer__max_features': [128,256,512,1024],
                 'classifier__C': [0.01, 0.1, 1, 10]}

svm_param_grid = {'vectorizer__max_features': [128,256,512,1024],
                  'classifier__C': [0.01, 0.1, 1, 10]}

rf_param_grid = {'vectorizer__max_features': [128,256,512,1024],
                 'classifier__criterion': ['gini', 'entropy'],
                 'classifier__n_estimators': [10,50,100]}
```

Para la búsqueda de hiperparámetros, emplearemos la validación cruzada K-fold con K=5. Esta técnica nos permite obtener una estimación más robusta del rendimiento del modelo en comparación a utilizar un único conjunto de validación. Al promediar los resultados de 5 evaluaciones diferentes, reducimos el riesgo de que el modelo se ajuste demasiado a un subconjunto específico de los datos.

Un aspecto fundamental para la evaluación objetiva de modelos de clasificación es la selección de una métrica de desempeño adecuada. En nuestro caso, al tratarse de un problema de clasificación multiclase con categorías no balanceadas, la elección del **macro-F1 score** se justifica por su capacidad para ponderar por igual la importancia de todas las clases, independientemente de su frecuencia en el conjunto de datos.

El macro-F1 score se calcula como la media aritmética de los índices F1 individuales obtenidos para cada clase. Esta métrica es especialmente útil en escenarios de clases no balanceadas, ya que no se ve influenciada por el soporte de cada categoría. Al calcular el F1 score para cada clase de manera independiente, se asegura que todas las clases contribuyan por igual a la métrica global, evitando que las clases mayoritarias dominen la evaluación y oculten el rendimiento del modelo en las clases minoritarias.


Este es un proceso exahustivo que puede taradar varios minutos (1 hora aproximadamente sin GPU y 30 minutos con GPU)

```python
print('Ejecutando busqueda de hiperparametros de la regresión logistica junto a TF_IDF ...')
lr_TFIDF_grid_search = GridSearchCV(pipeline_lr_TFIDF, lr_param_grid, cv=5, scoring='f1_macro', verbose=1, n_jobs=-1)
lr_TFIDF_grid_search.fit(X_train, y_train)
print()

print('Ejecutando busqueda de hiperparametros de la maquina de soporte vectorial junto a TF_IDF ...')
svm_TFIDF_grid_search = GridSearchCV(pipeline_svm_TFIDF, svm_param_grid, cv=5, scoring='f1_macro', verbose=1, n_jobs=-1)
svm_TFIDF_grid_search.fit(X_train, y_train)
print()

print('Ejecutando busqueda de hiperparametros del bosque aleatorio junto a TF_IDF ...')
rf_TFIDF_grid_search = GridSearchCV(pipeline_rf_TFIDF, rf_param_grid, cv=5, scoring='f1_macro', verbose=1, n_jobs=-1)
rf_TFIDF_grid_search.fit(X_train, y_train)
print()

print('Ejecutando busqueda de hiperparametros de la regresión logistica junto a Word2Vec ...')
lr_W2V_grid_search = GridSearchCV(pipeline_lr_W2V, lr_param_grid, cv=5, scoring='f1_macro', verbose=1, n_jobs=-1)
lr_W2V_grid_search.fit(X_train, y_train)
print()

print('Ejecutando busqueda de hiperparametros de la maquina de soporte vectorial junto a Word2Vec ...')
svm_W2V_grid_search = GridSearchCV(pipeline_svm_W2V, svm_param_grid, cv=5, scoring='f1_macro', verbose=1, n_jobs=-1)
svm_W2V_grid_search.fit(X_train, y_train)
print()

print('Ejecutando busqueda de hiperparametros del bsoque aleatorio junto a Word2Vec ...')
rf_W2V_grid_search = GridSearchCV(pipeline_rf_W2V, rf_param_grid, cv=5, scoring='f1_macro', verbose=1, n_jobs=-1)
rf_W2V_grid_search.fit(X_train, y_train)
print()
```
