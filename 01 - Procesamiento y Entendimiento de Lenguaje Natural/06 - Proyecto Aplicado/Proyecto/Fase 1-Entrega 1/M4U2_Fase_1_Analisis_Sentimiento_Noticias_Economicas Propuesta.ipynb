## **1.1. Introducción**
---

El crecimiento de los medios digitales ha generado un volumen cada vez mayor de información relacionada con acontecimientos económicos, mercados financieros, empresas e inversiones [1]. Las noticias financieras constituyen una fuente importante de información para diferentes actores económicos, ya que permiten conocer acontecimientos que pueden influir en la percepción y valoración de empresas, sectores y mercados [2]. Sin embargo, el gran volumen de información disponible dificulta su análisis manual [3]y hace necesario desarrollar mecanismos automáticos capaces de procesar y categorizar este tipo de contenido.

Dentro del campo del Procesamiento de Lenguaje Natural (NLP) [4], el análisis de sentimiento permite identificar la orientación expresada en un texto y clasificarla [5], generalmente, en categorías como positiva, negativa o neutral. Esta tarea adquiere características particulares en el ámbito financiero, debido a la presencia de terminología especializada, expresiones propias del lenguaje económico y relaciones semánticas cuyo significado depende del contexto.

Por ejemplo, expresiones como "superó las expectativas", "reducción de pérdidas", "caída de ingresos" o "revisión a la baja" requieren una interpretación contextual para determinar correctamente su orientación. Por esta razón, el análisis automático de sentimiento financiero representa un problema que requiere técnicas capaces de capturar las características particulares del lenguaje utilizado en este dominio.

En este contexto, el presente proyecto tiene como propósito desarrollar un algoritmo de Procesamiento de Lenguaje Natural orientado a la clasificación automática del sentimiento de noticias financieras escritas en español. Para ello, se utilizará el subconjunto en español del corpus Multilingual Financial Sentiment de *Hugging Face*, el cual contiene textos financieros etiquetados en tres categorías: positiva, neutral y negativa.

El algoritmo propuesto contemplará las diferentes etapas necesarias para transformar un texto de noticias financiers en una clasificación de sentimiento, incluyendo la preparación de los datos, procesamiento del lenguaje, representación de los textos, extracción de características, entrenamiento del modelo y evaluación de los resultados.

El proyecto busca desarrollar una solución especializada en el dominio financiero que permita identificar automáticamente la orientación de noticias económicas en español y analizar las características lingüísticas que influyen en la clasificación. Como resultado, se espera obtener un algoritmo capaz de procesar nuevos textos financieros y asignarlos automáticamente a una de las tres categorías de sentimiento definidas por el corpus.

El Corpus seleccionado para realizar el proyecto será el subconjunto datos en español del dataset *Multilingual Financial Sentiment* disponible en *Hugging Face*, el cual es un conjunto de datos de 39.829 oraciones de noticias financieras, anotadas con etiquetas de sentimiento en 7 idiomas y recopiladas de más de 80 fuentes de noticias financieras de todo el mundo.

Cada registro del corpus contiene un texto financiero y una etiqueta de sentimiento asociada. Las categorías disponibles son:

* Positive: textos cuya orientación financiera es positiva.
* Neutral: textos que no presentan una orientación positiva o negativa claramente predominante.
* Negative: textos cuya orientación financiera es negativa.

Para el desarrollo del algoritmo se utilizará exclusivamente la información correspondiente a los textos en español los cuales son 7,125.
